# DAILY TREND - TOTAL ORDERS

```python
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

df['day_name'] = df['order_date'].dt.day_name()

weekday_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

df['day_name'] = pd.Categorical(df['day_name'], categories=weekday_order, ordered=True)

orders_by_day = df.groupby('day_name', observed=False)['order_id'].nunique()

ax = orders_by_day.plot(kind='bar', figsize=(8,5), color='green', edgecolor='black')

plt.title("Total Orders by Day of Week")

plt.xlabel("Day of Week")

plt.ylabel("Number of Orders")

plt.xticks(rotation=45)

for i, val in enumerate(orders_by_day):
    plt.text(i, val + 20, str(val), ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()

plt.show()
```

## Creating a DataFrame from scratch (using a dictionary) - FOR JOINS EXERCISE

```python
import pandas as pd

data = {'pizza_name': ['The Spinach and Feta Pizza', 'The Classic Deluxe Pizza', 'The Pepperoni Pizza'],
        'pizza_size': ['L', 'M', 'S'],
        'unit_price': [12.75, 16.00, 12.00]}

df_pizzas = pd.DataFrame(data)

order_data = {
    'unit_price': [12.75, 16.00, 12.00],
    'total_price': [12.75, 16.00, 24.00],
    'pizza_size': ['L', 'M', 'S'],  # Shortened from 5 items to 3
    'order_id': [1, 2, 3],
    'pizza_name': ['The Spinach and Feta Pizza', 'The Classic Deluxe Pizza', 'The Pepperoni Pizza']
}

df_orders = pd.DataFrame(order_data)

print("DataFrame df_orders created successfully:")
print(df_orders) 
```

### INNER JOIN - 2 DICTIONARIES

```python
df_merged = pd.merge(df_orders, df_pizzas, 
                     on=['pizza_name', 'pizza_size'], 
                     how='inner', 
                     suffixes=('_order', '_menu'))

print("\nMerged DataFrame:")
print(df_merged)
```

### LEFT JOIN - 2 DICTIONARIES

```python
df_left_merged = pd.merge(df_orders, df_pizzas, 
                          on=['pizza_name', 'pizza_size'], 
                          how='left',  # <-- Changed from 'inner' to 'left'
                          suffixes=('_order', '_menu'))

print("Left Merged DataFrame:")
print(df_left_merged)
```

### RIGHT JOIN - 2 DICTIONARIES

```python
df_right_merged = pd.merge(df_orders, df_pizzas, 
                           on=['pizza_name', 'pizza_size'], 
                           how='right', # <--- Changed to 'right'
                           suffixes=('_order', '_menu'))

print("DataFrame after Right Join:")
print(df_right_merged)
```

### OUTER JOIN - 2 DICTIONARIES

```python
df_outer_merged = pd.merge(df_orders, df_pizzas, 
                          on=['pizza_name', 'pizza_size'], 
                          how='outer',  # <-- Changed to 'outer' for a full join
                          suffixes=('_order', '_menu'),
                          indicator=True) 
# Adding indicator=True helps visualize where the match came from
print("Full Outer Merged DataFrame:")
print(df_outer_merged)    
```

### CROSS JOIN - DATA FRAMES DEFINED FOR IT

```python
df_sizes = pd.DataFrame({'size_id': [1, 2], 'pizza_size': ['S', 'M']})
df_types = pd.DataFrame({'type_id': [101, 102], 'pizza_type': ['Pepperoni', 'Cheese']})
```

```python
df_cross_join = pd.merge(df_sizes, df_types, how='cross')

print("Cross Join Result (Cartesian Product):")
print(df_cross_join)
```

### SEMI JOIN - FOR DF ORDERS AND DF PIZZAS

```python
keys_to_keep = df_pizzas[['pizza_name', 'pizza_size']]

df_semi_join = df_orders[df_orders[['pizza_name', 'pizza_size']].isin(keys_to_keep).all(axis=1)]

print("\nSemi Join Result (Orders that have a menu match):")
print(df_semi_join)
```

### ANTI-JOIN - REVERSE OF SEMI JOIN

```python
keys_to_exclude = df_pizzas[['pizza_name', 'pizza_size']]

df_anti_join = df_orders[~df_orders[['pizza_name', 'pizza_size']].isin(keys_to_exclude).all(axis=1)]

print("\nAnti Join Result (Orders that do NOT have a menu match):")
print(df_anti_join)
```

### Merge As Of Table (pd.merge_asof())

```python
quotes = pd.DataFrame({
    'time': pd.to_datetime(['2025-01-01 09:30', '2025-01-01 09:45', '2025-01-01 10:00']),
    'price': [100, 101, 102]
}).sort_values('time')

trades = pd.DataFrame({
    'time': pd.to_datetime(['2025-01-01 09:32', '2025-01-01 09:55']),
    'volume': [5, 10]
}).sort_values('time')

# Setup data for an asof merge (must be sorted!)
```

```python
df_asof_merge = pd.merge_asof(trades, quotes, on='time', direction='backward')

print("\nMerge Asof Result:")
print(df_asof_merge)
# Merge trades with the 'as of' price available at that exact time or immediately before
```

### NEW DATAFRAMES - FOR TRYING JOINS EXERCISE

```python
import pandas as pd

data = {'pizza_id': ['1', '16736', '27419'],
        'order_id': ['1', '7367', '12063'],
        'pizza_name_id': ['hawaiian_m', 'pepperoni_l', 'green_garden_s'],
        'quantity': ['1','1','1']}
df_dominos = pd.DataFrame(data)

order_data = {'pizza_category':['Classic','Classic','Veggie'],
              'pizza_ingredients':['Iced Ham, Pineapple, Mozzarella Cheese','Mozzarella Cheese, Pepperoni','Spinach, Mushrooms, Tomatoes, Green Olives, Feta Cheese'],
              'pizza_name':['The Hawaiian Pizza','The Pepperoni Pizza','The Green Garden Pizza'] 
              }
df_category = pd.DataFrame(order_data)

name_map = {
    'hawaiian_m': 'The Hawaiian Pizza',
    'pepperoni_l': 'The Pepperoni Pizza',
    'green_garden_s': 'The Green Garden Pizza'
}
df_dominos['pizza_name'] = df_dominos['pizza_name_id'].map(name_map)


print("DataFrame df_orders created successfully:")
print(df_dominos) 
print(df_category)

```

## LEFT OUTER JOIN FOR DF DOMINOS AND DF CATEGORY DATA FRAMES

```python
left_join_df = pd.merge(left=df_dominos, 
                        right=df_category, 
                        on='pizza_name', 
                        how='left')
print(left_join_df)
```

## RIGHT JOIN FOR DF DOMINOS AND DF CATEGORY DATA FRAMES

```python
right_join_df = pd.merge(left=df_dominos, 
                         right=df_category, 
                         on='pizza_name', 
                         how='right')
print(right_join_df)
```

## INNER JOIN FOR DF DOMINOS AND DF CATEGORY DATA FRAMES

```python
inner_join_df = pd.merge(left=df_dominos, 
                         right=df_category, 
                         on='pizza_name', 
                         how='inner')

print("\n--- INNER JOIN Result ---")
print(inner_join_df)
```

## FULL OUTER JOIN FOR DF DOMINOS AND DF CATEGORY DATA FRAMES

```python
outer_join_df = pd.merge(left=df_dominos, 
                         right=df_category, 
                         on='pizza_name', 
                         how='outer')

print("\n--- OUTER JOIN Result ---")
print(outer_join_df)
```

## CROSS JOIN FOR DF DOMINOS AND DF CATEGORY DATA FRAMES

```python
cross_join_df = pd.merge(left=df_dominos, 
                         right=df_category, 
                         how='cross')

print("\n--- CROSS JOIN Result ---")
print(cross_join_df)
```

## SEMI JOIN and ANTI JOIN (Implemented using isin / Merging)

```python
# Method 1: Using isin()
semi_join_df = df_dominos[df_dominos['pizza_name'].isin(df_category['pizza_name'])]

print("\n--- SEMI JOIN Result (columns only from left DF) ---")
print(semi_join_df)
```

## ANTI-JOIN

```python
# Method 1: Using isin() combined with negation (~)
anti_join_df = df_dominos[~df_dominos['pizza_name'].isin(df_category['pizza_name'])]

print("\n--- ANTI JOIN Result (rows in left DF that are NOT in right DF) ---")
print(anti_join_df)
```

## Merge As Of Table (pd.merge_asof())

```python
# Example for pd.merge_asof()
trade_data = pd.DataFrame({
    'time': pd.to_datetime(['2023-01-01 09:00', '2023-01-01 10:30', '2023-01-01 11:45']),
    'stock': ['GOOG', 'GOOG', 'GOOG'],
    'quantity': [100, 150, 80]
})

quote_data = pd.DataFrame({
    'time': pd.to_datetime(['2023-01-01 09:01', '2023-01-01 10:00', '2023-01-01 11:30', '2023-01-01 12:00']),
    'stock': ['GOOG', 'GOOG', 'GOOG', 'GOOG'],
    'price': [100.5, 101.2, 102.1, 103.5]
})

# Sort both dataframes by the key (time)
trade_data = trade_data.sort_values('time')
quote_data = quote_data.sort_values('time')

merge_asof_df = pd.merge_asof(trade_data, quote_data, 
                              on='time', 
                              by='stock',
                              direction='backward') # Use the last known price *as of* the trade time

print("\n--- MERGE ASOF Result (Time-Series Join) ---")
print(merge_asof_df)
```
